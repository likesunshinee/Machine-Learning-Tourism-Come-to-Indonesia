```python
def classify_entry(name):
    name = str(name).strip()

    # Sub-total berdasarkan jenis pintu
    if name in [
        "A. Pintu Udara",
        "B. Pintu Laut",
        "C. Pintu Darat"
    ]:
        return "Sub-Total"

    # Grand total
    elif "Total" in name or "Jumlah" in name:
        return "Grand Total"

    # Kategori gabungan / residual
    elif (
        name == "Lainnya"
        or name.endswith("Lainnya")
    ):
        return "Pintu Masuk Lainnya"

    # Pintu masuk individual
    else:
        return "Pintu Masuk"
```

# BUAT FUNGSI FINA

```python
def clean_tourism_data(file):
    
    # 1. Ambil tahun dari nama file
    year = int(file.split("_")[-1].replace(".csv", ""))
    
    # 2. Baca CSV sebagai data mentah
    df = pd.read_csv(file, header=None)
    
    # 3. Gunakan baris ke-4 sebagai header
    df.columns = df.iloc[3]
    
    # 4. Buang baris metadata
    df = df.iloc[4:].copy()
    
    # 5. Beri nama kolom pertama
    df.columns.values[0] = "Pintu Masuk"
    
    # 6. Ambil hanya kolom yang diperlukan
    df = df[["Pintu Masuk"] + months]
    
    # 7. Ubah wide → long
    df = df.melt(
        id_vars=["Pintu Masuk"],
        var_name="Bulan",
        value_name="Jumlah"
    )
    
    # 8. Bersihkan nilai "-"
    df["Jumlah"] = df["Jumlah"].replace("-", np.nan)
    
    # 9. Ubah Jumlah menjadi numerik
    df["Jumlah"] = pd.to_numeric(
        df["Jumlah"],
        errors="coerce"
    )
    
    # 10. Klasifikasi jenis data
    df["Jenis Data"] = df["Pintu Masuk"].apply(
        classify_entry
    )
    
    # 11. Tambahkan tahun
    df["Tahun"] = year
    
    # 12. Konversi nama bulan → nomor
    df["Bulan_Num"] = df["Bulan"].map(month_map)
    
    # 13. Buat tanggal
    df["Tanggal"] = pd.to_datetime(
        dict(
            year=df["Tahun"],
            month=df["Bulan_Num"],
            day=1
        )
    )
    
    # 14. Urutkan kolom
    df = df[
        [
            "Pintu Masuk",
            "Bulan",
            "Jumlah",
            "Jenis Data",
            "Tahun",
            "Bulan_Num",
            "Tanggal"
        ]
    ]
    
    return df
```

# EDA SEASONALITY

```python
seasonal_normal = final_df[
    (final_df["Jenis Data"].isin([
        "Pintu Masuk",
        "Pintu Masuk Lainnya"
    ]))
    &
    (final_df["Tahun"].between(2016, 2019))
].copy()

seasonality = (
    seasonal_normal
    .groupby(["Bulan", "Bulan_Num"])["Jumlah"]
    .sum(min_count=1)
    .reset_index()
)
```

```python
monthly_normal = (
    seasonal_normal
    .groupby(["Tahun", "Bulan_Num", "Bulan"])["Jumlah"]
    .sum(min_count=1)
    .reset_index()
)

seasonality = (
    monthly_normal
    .groupby(["Bulan_Num", "Bulan"])["Jumlah"]
    .mean()
    .reset_index()
    .sort_values("Bulan_Num")
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    seasonality["Bulan"],
    seasonality["Jumlah"],
    marker="o"
)

plt.title("Pola Musiman Kunjungan Wisatawan Mancanegara (2016–2019)")
plt.xlabel("Bulan")
plt.ylabel("Rata-rata Kunjungan")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
pandemic_change = (
    (pandemic_analysis.loc[2021] -
     pandemic_analysis.loc[2019])
    / pandemic_analysis.loc[2019]
) * 100

print("Perubahan 2019 → 2021 (%):")
print(pandemic_change)
```

```python
recovery_plot = recovery_2025.sort_values()

plt.figure(figsize=(10, 5))

recovery_plot.plot(
    kind="barh"
)

plt.axvline(
    100,
    linestyle="--"
)

plt.title("Recovery Kunjungan Wisatawan Tahun 2025 terhadap Baseline 2019")
plt.xlabel("Recovery (%)")
plt.ylabel("Jalur Masuk")

plt.show()
```

```python
entry_recovery = final_df[
    (final_df["Jenis Data"] == "Pintu Masuk") &
    (final_df["Tahun"].isin([2019, 2021, 2025]))
].copy()

entry_recovery_total = (
    entry_recovery
    .groupby(["Pintu Masuk", "Tahun"])["Jumlah"]
    .sum(min_count=1)
    .reset_index()
)

entry_recovery_pivot = entry_recovery_total.pivot(
    index="Pintu Masuk",
    columns="Tahun",
    values="Jumlah"
)

entry_recovery_pivot["Recovery_2025"] = (
    entry_recovery_pivot[2025]
    / entry_recovery_pivot[2019]
) * 100

entry_recovery_sorted = (
    entry_recovery_pivot
    .sort_values("Recovery_2025", ascending=False)
)

print(
    entry_recovery_sorted[
        [2019, 2021, 2025, "Recovery_2025"]
    ]
)
```

```python
recovery_entry_plot = (
    entry_recovery_sorted["Recovery_2025"]
    .sort_values()
)

plt.figure(figsize=(10, 8))

recovery_entry_plot.plot(
    kind="barh"
)

plt.axvline(
    100,
    linestyle="--"
)

plt.title(
    "Recovery Pintu Masuk 2025 terhadap Baseline 2019"
)

plt.xlabel("Recovery (%)")
plt.ylabel("Pintu Masuk")

plt.show()
```
